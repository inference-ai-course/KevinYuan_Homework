```python
from unsloth import FastLanguageModel
from trl import SFTTrainer
from transformers import TrainingArguments
from datasets import load_dataset

# Load the base LLaMA 3 7B model in 4-bit mode (dynamic 4-bit quantization)
model_name = "unsloth/Meta-Llama-3.1-8B-Instruct-bnb-4bit"
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=model_name,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)

# Add LoRA adapters
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj",
                    "gate_proj", "up_proj", "down_proj"],
    lora_alpha=16,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=3407,
)

# Load our synthetic Q&A dataset
dataset = load_dataset("json", data_files="synthetic_qa.jsonl", split="train")

# Initialize the trainer for Supervised Fine-Tuning (SFT)
trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=2048,
    args=TrainingArguments(
        output_dir="llama3-7b-qlora-finetuned",
        per_device_train_batch_size=4,   # small batch size for Colab GPU
        gradient_accumulation_steps=4,   # accumulate gradients to simulate larger batch
        num_train_epochs=2,
        learning_rate=2e-4,
        bf16=True,  
        logging_steps=50,
        save_strategy="epoch"
    )
)

trainer.train()
model.save_pretrained("llama3-7b-qlora-finetuned")
```

```python
test_questions = [
    "What are the main challenges in aligning large language models with human values?",
    "How do researchers measure the effectiveness of safety mechanisms in AI systems?",
    "What role does model size play in the reliability of LLM outputs?",
    "Can you explain the relationship between prompt engineering and model robustness?",
    "What are some common techniques for preventing adversarial attacks on language models?",
    "How does the architecture of transformers affect their ability to handle long contexts?",
    "What is the difference between instruction-tuning and alignment training?",
    "How do researchers evaluate whether an LLM has truly learned from fine-tuning?",
    "What are the trade-offs between model performance and computational efficiency?",
    "How might fine-tuning on domain-specific data affect a model's general capabilities?"
]

# Define system prompt
system_prompt = "You are a helpful academic Q&A assistant specialized in scholarly content."

# Load base model (without fine-tuning)
print("Loading base model...")
base_model, base_tokenizer = FastLanguageModel.from_pretrained(
    model_name=model_name,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)
FastLanguageModel.for_inference(base_model)

# Load fine-tuned model
print("Loading fine-tuned model...")
ft_model, ft_tokenizer = FastLanguageModel.from_pretrained(
    model_name="llama3-7b-qlora-finetuned",
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
)
FastLanguageModel.for_inference(ft_model)

print("\nComparing Base vs Fine-Tuned Model...\n")
print("=" * 80)

for i, q in enumerate(test_questions, 1):
    prompt_input = f"<|system|>{system_prompt}<|user|>{q}<|assistant|>"
    
    # Tokenize input and generate output with each model
    base_input_ids = base_tokenizer(prompt_input, return_tensors='pt').input_ids.cuda()
    ft_input_ids = ft_tokenizer(prompt_input, return_tensors='pt').input_ids.cuda()
    
    base_output_ids = base_model.generate(base_input_ids, max_new_tokens=150, temperature=0.7, do_sample=True)
    ft_output_ids = ft_model.generate(ft_input_ids, max_new_tokens=150, temperature=0.7, do_sample=True)
    
    # Decode the outputs
    base_answer = base_tokenizer.decode(base_output_ids[0], skip_special_tokens=True)
    ft_answer = ft_tokenizer.decode(ft_output_ids[0], skip_special_tokens=True)
    
    # Extract only the assistant's response
    base_answer = base_answer.split('<|assistant|>')[-1].strip()
    ft_answer = ft_answer.split('<|assistant|>')[-1].strip()
    
    print(f"\n[Question {i}]: {q}")
    print(f"\n[Base Model]: {base_answer}")
    print(f"\n[Fine-Tuned Model]: {ft_answer}")
    print("-" * 80)
```
